```python
import pandas as pd
print("Pandas is ready to work!")
```

```python
# 1. Create a simulated messy transaction dataset
raw_data = {
    'TransactionID': ['TXN001', 'TXN002', 'TXN003', 'TXN004', 'TXN004', 'TXN005', 'TXN006'],
    'InvoiceNum': ['INV-1024', None, 'INV-1026', 'INV-1027', 'INV-1027', 'INV-1028', 'INV-1029'],
    'Vendor': ['Apex Logistics', 'Chevron Corp', 'Global Tech', 'Office Depot', 'Office Depot', 'Acme Corp', 'Industrial Supplies'],
    'Amount': [4500.00, 12500.50, None, 250.00, 250.00, 18500.00, 9400.00],
    'Status': ['Paid', 'Pending', 'Paid', 'Paid', 'Paid', 'Pending', 'Paid']
}

# 2. Turn it into a Pandas DataFrame and save it as a CSV file
df_raw = pd.DataFrame(raw_data)
df_raw.to_csv('raw_ledger.csv', index=False)

print("Messy ledger file 'raw_ledger.csv' created successfully!")
print(df_raw)
```

```python
# 1. Load the raw ledger file
df = pd.read_csv('raw_ledger.csv')
print("--- STARTING FINANCIAL LEDGER AUDIT ---")
print(f"Total raw records loaded: {len(df)}")

# 2. Drop duplicate rows
df = df.drop_duplicates()

# 3. Drop rows where the Amount is missing entirely
df = df.dropna(subset=['Amount'])

# 4. Fill missing invoice numbers with a standard placeholder
df['InvoiceNum'] = df['InvoiceNum'].fillna('UNKNOWN-CHECK')
print(f"Records remaining after cleaning: {len(df)}")

# 5. Isolate transactions exceeding $10,000 for manager/auditor review
high_risk_threshold = 10000.00
high_risk_df = df[df['Amount'] > high_risk_threshold]
print(f"AUDIT ALERT: Found {len(high_risk_df)} transaction(s) exceeding ${high_risk_threshold:,.2f}")

# 6. Save both reports to clean Excel files
df.to_excel('clean_master_ledger.xlsx', index=False)
high_risk_df.to_excel('high_risk_audit_report.xlsx', index=False)

print("\n--- AUDIT COMPLETE ---")
print("Clean Master Ledger:")
print(df)
print("\nHigh-Risk Audit Exception Report:")
print(high_risk_df)
```

```python
# 1. Calculate total spend and status by vendor from our clean data
vendor_summary = df.groupby(['Vendor', 'Status'])['Amount'].sum().reset_index()

# 2. Sort from highest spend to lowest
vendor_summary = vendor_summary.sort_values(by='Amount', ascending=False)

# 3. Format the numbers for executive presentation
print("--- VENDOR FINANCIAL SUMMARY ---")
print(vendor_summary)

# 4. Export to a third Excel summary sheet
vendor_summary.to_excel('vendor_spend_summary.xlsx', index=False)
print("\nExported: 'vendor_spend_summary.xlsx' is ready for leadership!")
```
